## Popcorn B

```python
# Prediction for n = 8:
# Linear worst case = 8
# Binary worst case = log2(8) + 1 = 3 + 1 = 4
# Expected Output:
# 8
# 8
# 4
# Prediction for n = 16:
# Linear worst case = 16 (doubles from 8)
# Binary worst case = log2(16) + 1 = 4 + 1 = 5 (increases by 1)
# Expected Output:
# 16
# 16
# 5

import math

n = 16  # Changed from 8 to 16 as required
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1

print("Catalog size (n): " + str(n))
print("Linear search worst-case checks: " + str(linear_checks))
print("Binary search worst-case checks: " + str(binary_checks))
```

## Homework

```python
"""

DESIGN THINKING SUMMARY:
- Empathize / Need: Librarian wants a simple tool to compare search work.
- Goal: Show clear search checks and pair counts without confusing code math.
- Methods: Compared Linear Search (scans one-by-one) vs Binary Search (repeatedly halves).
- Prototype: Python script computing visits (n) and pair combinations (n * n).
- Tests and Revision: Initial output was just raw numbers (12, 12, 144). I added descriptive labels to every print line.

"""

# PREDICTION for n = 12:
# Books = 12, Visits = 12, Pairs = 12 * 12 = 144
# Expected Output: 12, 12, 144

# PREDICTION for n = 24 (Doubled Catalog):
# Books = 24, Visits = 24, Pairs = 24 * 24 = 576
# Expected Output: 24, 24, 576

def analyze_catalog_growth(n):
    visits = n
    pairs = n * n
    
    print("--- Catalog Analysis (n = {}) ---".format(n))
    print("Total Books in Catalog: {}".format(n))
    print("Visits (Linear Scan): {}".format(visits))
    print("Ordered Pairs (All-Pair Comparisons): {}\n".format(pairs))

# Main Run (n = 12)
analyze_catalog_growth(12)

# Doubled Run (n = 24)
analyze_catalog_growth(24)
```
