## Popcorn Hack

1. The **checkbox** on a toggle keeps track of whether it is on or off.
2. The **slider** is the visible part of the toggle that changes color when it is turned on.
3. The .switch class controls the toggles position, size, padding, and **alignment**.
4. The **border-radius** Property makes the slider have rounded edges.
5. The &before class dictates the **white circle** of the toggle.

## Fix the Bug

```python
%%html
<!-- UI_RUNNER: Toggles Popcorn - find the bug -->
<label class="ocs__toggle ocs__toggle--green">
  <input id="bugfix-toggle" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Enable Notifications</span>
</label>
<p id="bugfix-result"></p>
<script>
(function () {
  // Same check the lesson's runner does: is the track class present now?
  const ok = document.querySelector("#bugfix-toggle + .ocs__toggle-track") !== null;
  document.getElementById("bugfix-result").textContent = ok ? "Toggle works! 🎉" : "Incorrect, almost there!";
})();
</script>
```

## Homework Hack

```python
from IPython.display import HTML, display

# Define outputElement so Python doesn't return a NameError in Jupyter
class MockElement:
    def __init__(self):
        self.innerHTML = ""

outputElement = MockElement()

outputElement.innerHTML = ''.join([
## .checked was interesting because it lets code instantly tell whether a toggle is currently on or off!
  '<div>',
  '  <h3>My Interactive Toggles Homework</h3>',
  '',
  '  <!-- Toggle 1: Notifications -->',
  '  <label class="ocs__toggle ocs__toggle--green" style="margin-bottom: 10px; display: block;">',
  '    <input id="toggle-1" class="ocs__toggle-input" type="checkbox">',
  '    <span class="ocs__toggle-track" aria-hidden="true"></span>',
  '    <span class="ocs__toggle-label">Enable Push Notifications</span>',
  '  </label>',
  '  <p id="msg-1" hidden>🔔 Notifications are turned on!</p>',
  '',
  '  <!-- Toggle 2: Dark Mode -->',
  '  <label class="ocs__toggle ocs__toggle--green" style="margin-bottom: 10px; display: block;">',
  '    <input id="toggle-2" class="ocs__toggle-input" type="checkbox">',
  '    <span class="ocs__toggle-track" aria-hidden="true"></span>',
  '    <span class="ocs__toggle-label">Activate Dark Mode</span>',
  '  </label>',
  '  <p id="msg-2" hidden>🌙 Dark mode theme is active!</p>',
  '',
  '  <!-- Toggle 3: Privacy Mode -->',
  '  <label class="ocs__toggle ocs__toggle--green" style="margin-bottom: 10px; display: block;">',
  '    <input id="toggle-3" class="ocs__toggle-input" type="checkbox">',
  '    <span class="ocs__toggle-track" aria-hidden="true"></span>',
  '    <span class="ocs__toggle-label">Strict Privacy Mode</span>',
  '  </label>',
  '  <p id="msg-3" hidden>🔒 Privacy mode is protecting your data!</p>',
  '',
  '  <hr style="margin: 15px 0;">',
  '  <p id="toggle-counter" style="font-weight: 600;">Total Active Toggles: 0</p>',
  '</div>',
  '',
  '<script>',
## querySelectorAll is useful because it selects all toggle inputs so we can loop through them easily and track active states.',
  '  const inputs = document.querySelectorAll(".ocs__toggle-input");',
  '  const counterDisplay = document.getElementById("toggle-counter");',
  '',
  '  function updateToggles() {',
  '    let activeTotal = 0;',
  '    inputs.forEach(function(input) {',
  '      if (input.id === "toggle-1") {',
  '        document.getElementById("msg-1").hidden = !input.checked;',
  '      } else if (input.id === "toggle-2") {',
  '        document.getElementById("msg-2").hidden = !input.checked;',
  '      } else if (input.id === "toggle-3") {',
  '        document.getElementById("msg-3").hidden = !input.checked;',
  '      }',
  '      if (input.checked) {',
  '        activeTotal++;',
  '      }',
  '    });',
## Comment 2: Updating the counter text dynamically on every change will return real-time feedback for the user.
  '    counterDisplay.textContent = "Total Active Toggles: " + activeTotal;',
  '  }',
  '',
  '  inputs.forEach(function(input) {',
  '    input.addEventListener("change", updateToggles);',
  '  });',
  '',
  '  updateToggles();',
  '</script>'
])

# Render the HTML output in Jupyter
display(HTML(outputElement.innerHTML))
```
