## Homework Hack

```python
message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# TODO 1: add to risk_score using conditionals, one signal at a time
# Signal 1: +4 if the message mentions gift cards
# We don't need an 'else' because we only accumulate points when a risk signal is present; if absent, we do nothing.
if "gift card" in message.lower():
    risk_score += 4

# Signal 2: +3 if the amount asked for is over 100
# We don't need an 'else' because absent signals should not alter or reset the score.
if amount_requested > 100:
    risk_score += 3

# Signal 3: +2 if the sender is not a known contact
# We don't need an 'else' because signals operate independently and non-triggering signals require no score adjustment.
if not known_contact:
    risk_score += 2

# TODO 2: report on the final score with an if / elif / else chain
# TODO 3: count comment in this form, with your real numbers:
# 3 thresholds -> 4 outcomes -> 4 answers

if risk_score >= 7:
    print(f"Risk Score: {risk_score} - Critical Risk: High likelihood of a scam. Do not engage or send money.")
elif risk_score >= 4:
    print(f"Risk Score: {risk_score} - Moderate Risk: Exercise caution and verify the sender's identity.")
elif risk_score >= 1:
    print(f"Risk Score: {risk_score} - Low Risk: Minor signals detected, double check message details.")
else:
    print(f"Risk Score: {risk_score} - Safe: No risk signals detected.")
```
