## Popcorn 1

```python
def report_status(location, status):
    # Print a status update using both parameters
    print("Status update from " + location + ": " + status)

# Calling report_status() two times
report_status("Poway", "All clear")
report_status("Rancho Bernardo", "Power restored")
```

## Popcorn 2

```python
def is_priority(incident_type):
    # Returns True if incident is a "Medical Emergency"
    return incident_type == "Medical Emergency"

def process_report(location, incident_type):
    priority = is_priority(incident_type)
    if priority:
        print("PRIORITY: " + incident_type + " reported in " + location)
    else:
        print("STANDARD REPORT: " + incident_type + " reported in " + location)

# Testing procedure calls
process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
```

## Homework Hack

```python
# Helper Procedure
def check_priority(incident_type):
    # Returns True for urgent priority incidents
    return incident_type == "Medical Emergency" or incident_type == "Structure Fire"

# Outer Procedure
def dispatch_response(location, incident_type):
    # Call check_priority helper function
    is_urgent = check_priority(incident_type)
    
    if is_urgent:
        print("IMMEDIATE DISPATCH: Sending emergency units to " + location + " for " + incident_type)
    else:
        print("ROUTINE DISPATCH: Logging report for " + incident_type + " at " + location)

# Calling dispatch_response at least three times
dispatch_response("Poway", "Medical Emergency")
dispatch_response("Sabre Springs", "Structure Fire")
dispatch_response("Rancho Peñasquitos", "Traffic Jam")
```
