## Homework Hack

```python
%%html

<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <h3 class="ocs_lead">Have you ever wanted to...</h3>
  <p class="ocs_text">pat a cat? Well good for you we will give step by step instructions on how to pat a cat!</p>
  <p class="ocs_text"><strong>How to pat a cat:</strong></p>
  <ol>
    <li>Put out your hand</li>
    <li>Touch the cat</li>
    <li>Move your hand back and forth</li>
  </ol>
  <h3 class="ocs_section-title">Congrats! You pat a cat :D</h3>
</div>
```

The %%html and the UI_RUNNER are present. All the formatting is correct; there is a heading, paragraph and ocs_ classes that are applied.
