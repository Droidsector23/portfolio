```python
seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]   # use real class data

frequency = {}
for season in seasons:
    if season in frequency:
        frequency[season] += 1
    else:
        frequency[season] = 1


# 1. Build a list of seasons with 2 or more votes
popular = []
for season, count in frequency.items():
    if count >= 2:
        popular.append(season)

# 2. Print that list
# TODO: print(popular)
print(popular)
```

```python
votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips",
         "pretzels", "cookies", "chips", "fruit", "chips", "cookies"]

prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# LESSON 1 - indexing
# 1. Print how many votes were cast
print(len(votes))
# 2. Print the last vote using indexing
print(votes[-1])

# LESSON 2 - sum and average
# 3. Use a loop to total prices (do NOT use sum())
total = 0
for price in prices:
    total += price

# 4. Print "Total: <total>" and "Average: <average>" as integers using the int() function
#ex: print(int(total)) would print the integer value of total
print("Total:", int(total))
print("Average:", int(total/len(prices)))

# LESSON 3 - count and filter
# 5. Build a frequency dictionary of votes and print it
frequency = {}
for vote in votes:
    if vote in frequency:
        frequency[vote] += 1
    else:
        frequency[vote] = 1
print(frequency)

# 6. Build a list of snacks with 3 or more votes and print it
popular = []
for snack, count in frequency.items():
    if count >= 3:
        popular.append(snack)
print(popular)
```
